# 03 — Train Models

Trains Decision Tree, Naive Bayes, k-NN, Random Forest, and AdaBoost classifiers, each on both the original (imbalanced) and SMOTE-balanced training sets — ten model configurations total. Requires `02_preprocessing.ipynb` to have been run first.

```python
import pickle

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier

from config import DATA_DIR, RANDOM_STATE

MODELS_DIR = DATA_DIR / "models"
MODELS_DIR.mkdir(parents=True, exist_ok=True)
```

## Load train splits

```python
X_train = pd.read_csv(DATA_DIR / "X_train.csv")
y_train = pd.read_csv(DATA_DIR / "y_train.csv").iloc[:, 0]
X_train_bal = pd.read_csv(DATA_DIR / "X_train_balanced.csv")
y_train_bal = pd.read_csv(DATA_DIR / "y_train_balanced.csv").iloc[:, 0]

print(f"Original train shape: {X_train.shape}")
print(f"SMOTE-balanced train shape: {X_train_bal.shape}")
```

## Model definitions

Five classifiers as required by the course guidelines: Decision Tree, Naive Bayes, k-NN, and two ensemble methods (Random Forest, AdaBoost). Hyperparameters are simple, fixed choices (not tuned via grid search) — justify these choices in the report rather than in code comments.

```python
def build_models():
    return {
        "decision_tree": DecisionTreeClassifier(max_depth=8, random_state=RANDOM_STATE),
        "naive_bayes": GaussianNB(),
        "knn": KNeighborsClassifier(n_neighbors=15),
        "random_forest": RandomForestClassifier(
            n_estimators=200, max_depth=12, random_state=RANDOM_STATE, n_jobs=-1
        ),
        "adaboost": AdaBoostClassifier(n_estimators=100, random_state=RANDOM_STATE),
    }


def train_and_save(X, y, suffix):
    models = build_models()
    for name, model in models.items():
        model.fit(X, y)
        out_path = MODELS_DIR / f"{name}_{suffix}.pkl"
        with open(out_path, "wb") as f:
            pickle.dump(model, f)
        print(f"Trained and saved: {out_path.name}")
```

## Train on the original (imbalanced) training set

```python
train_and_save(X_train, y_train, "orig")
```

## Train on the SMOTE-balanced training set

```python
train_and_save(X_train_bal, y_train_bal, "smote")

print(f"\nAll models saved to {MODELS_DIR}")
```
